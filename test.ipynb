```python
# visualize_graph.py
from skills.orchestrator import build_graph
from IPython.display import Image, display

# Build the graph
graph = build_graph()
print("✅ Graph compiled successfully!")

# Option 1: Display as Mermaid text (works without extra packages)
mermaid_code = graph.get_graph().draw_mermaid()
print("\n📊 Mermaid Diagram (copy into https://mermaid.live/):")
print(mermaid_code)

# Option 2: Display as PNG (requires pygraphviz)
try:
    import pygraphviz  # noqa: F401 – check if installed
    png_data = graph.get_graph().draw_mermaid_png()
    display(Image(png_data))
    print("✅ PNG diagram displayed above.")
except ImportError:
    print("\n⚠️ pygraphviz not installed. Install it with:")
    print("   pip install pygraphviz")
    print("Then re‑run to see the PNG diagram.")
```
